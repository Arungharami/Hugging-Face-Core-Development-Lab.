# Notebook 01: Hugging Face Hub Discovery & Inspection

This notebook demonstrates how to programmatically search, inspect, and audit Hugging Face models, datasets, and Spaces using `hf_core_lab`.

```python
import sys

sys.path.insert(0, '../src')

from hf_core_lab.config import LabConfig
from hf_core_lab.hub.discovery import HubDiscoveryEngine
from hf_core_lab.hub.validators import CardValidator
from hf_core_lab.models.reports import ReportGenerator
```

```python
# Initialize Discovery Engine
config = LabConfig()
engine = HubDiscoveryEngine(config=config)

# Search Text Classification Models
models = engine.search_models(query="text-classification", limit=5)
for m in models:
    print(f"ID: {m.model_id} | Downloads: {m.downloads:,} | License: {m.license}")
```

```python
# Validate Metadata Compliance
results = [CardValidator.validate_model_metadata(m) for m in models]
print(ReportGenerator.validation_to_markdown(results))
```
